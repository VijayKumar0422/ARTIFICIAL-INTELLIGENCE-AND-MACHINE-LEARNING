# Lamda_Functions

```python
# lambda function --> anonymous function
```

```python
# simple

def sq(x):
  return x*x

print(sq(5))
```

```python
# using lambda
```

```python
# syntax
# lambda parameter: expression

sq = lambda x: x*x
sq(6)
```

```python
# write a lambda function to return the twice of the given numbers

twice = lambda x: 2*x
twice(4)
```

```python
# write a coade to reurn the sum of two given numbers

add = lambda x,y: x+y
add(4,6)
```

```python
# write a coad to return the max of two numbers

# simple coad using function

# def mx(a,b):
#   if a > b:
#     return a
#   else:
#     return b

# max(4,9)

# using functional_programming

max = lambda a, b: a if a > b else b
max(5,7)
```

```python
# write a function to return if the given number is even or odd
```

```python
# simple function

def check(n):
  if n % 2 == 0:
    return "Even"
  else:
    return "Odd"

check(6)
```

```python
# using Functional_programming

check = lambda n: "Even" if n % 2 == 0 else "Odd"

check(8)
```

# map

```python
# map(function, iterable) --> it returns a map object --> convert it into list --> list(map)
```

```python
# write a function to return a list of the cube of every number in the given list.
```

```python
def cube(lst):
  cb = []
  for n in lst:
    cb.append(n**3)

  return cb

cube([5, 7, 3])
```

```python
def cb(n):
  return n**3

cube = []
lst = [4, 6, 2]
for elem in lst:
  cube.append(cb(elem))

print(cube)
```

```python
# using map

lst = [4, 6, 2]
cube = list(map(lambda x: x**3, lst))
print(cube)
```

```python
cube = list(map(lambda x: x**3, [4, 6, 2]))
print(cube)
```

```python
print(list(map(lambda x: x**3, [4, 6, 2])))
```

#Filter

```python
num = [1, 2, 3, 4, 5, 6, 7]

# write a coad to create a list of all the even numbers of the given list

even = []

for n in num:
  if n % 2 == 0:
    even.append(n)

print(even)
```

```python
num = [1, 2, 3, 4, 5, 6, 7]

even = list(filter(lambda x: x % 2 == 0, num))
print(even)
```

```python
even = list(filter(lambda x: x % 2 == 0, [1, 2, 3, 4, 5, 6, 7]))
print(even)
```

```python
odd = list(filter(lambda x: x % 2 != 0, [1, 2, 3, 4, 5, 6, 7]))
print(odd)
```
