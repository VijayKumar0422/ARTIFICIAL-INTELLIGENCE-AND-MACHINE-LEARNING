# catching specific errors

```python
# ye specific error bata deta hai

try:
  n = int(input())
  print(100/n)

except Exception as e:
  print(e)
```

```python
try:
  n = int(input())
  print(100/n)

except Exception as e:
  print(e)
```

```python
try:
  n = int(input())
  print(100/n)

except ZeroDivisionError:
  print("Division by 0 is not allow")
except ValueError:
  print("Please enter numbers only")
```

```python
try:
  n = int(input())
  print(100/n)
  print("Hello"

except ZeroDivisionError:
  print("Division by 0 is not allow")
except ValueError:
  print("Please enter numbers only")
except Exception as e:
  print("Some error has occured")
```

```python
try:
  n = int(input())
  print(100/n)
  a = [3, 6]
  print(a[4])
except ZeroDivisionError:
  print("Division by 0 is not allow")
except ValueError:
  print("Please enter numbers only")
except Exception as e:
  print("Some error has occured")
```

```python
# else block runs only when no exception occurs

try:
  num = int(input())

except ValueError:
  print("Invalid input")
else:
  print("You entered ", num)
```

```python
try:
  num = int(input())

except ValueError:
  print("Invalid input")
else:
  print("You entered ", num)
```

```python
# finally block always executes irrespective wheater there is an error or not

try:
  num = int(input())

except ValueError:
  print("Invalid input")
finally:
  print("Pogram finished")
```
