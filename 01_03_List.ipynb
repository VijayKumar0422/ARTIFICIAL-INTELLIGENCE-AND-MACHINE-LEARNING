```python
# write a coad to find the sum of all the elements of the list.

mark = [5, 4, 6, 9]
sm = 0
for elem in mark:
  sm = sm + elem

print(sm)
```

```python
# sum karne ka short way

mark = [5, 4, 6, 9]
print(sum(mark))
```

```python
# write a codo to find the average of all the elements of the list.

mark = [60, 70, 80, 90]
n = len(mark)
sm = 0
for elem in mark:
  sm = sm + elem

print(sm/n)

```
