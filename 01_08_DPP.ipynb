```python
# Given a string, write a code to conver it to lowrcase, U cannot use lower()

s = "Vijay"

new_s = ""

for ch in s:
  if ch <= 'Z' and ch >= 'A':
    new_s += chr(ord(ch) + 32)
  else:
    new_s += ch

print(new_s)
```

```python
# Given a string, write a code to conver it to uppercase., U cannot use lower()

s = "vijay"

new_s = ""

for ch in s:
  if ch <= 'z' and ch >= 'a':
    new_s += chr(ord(ch) - 32)
  else:
    new_s += ch

print(new_s)
```
