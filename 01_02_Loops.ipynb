# WHILE LOOP

```python
# 1. initialization
# 2. condition
# 3. updation
```

```python
# print all the int from 1 to 5.

i = 1

while i <= 5:
  print(i)
  i = i+1
```

```python
# write a while loop to print Hi 7 times on the screen


i = 1

while i <= 7:
  print(i, "Hi")
  i += 1  # i = i+1
```

```python
i = 7

while i >= 1:
  print(i, "Hi")
  i -= 1
```

```python
# print all the even numbers between 0 and 10 (both included) using a while loop

i = 0

while i <= 10:
  print(i)
  i += 2
```

```python
i = 0

while i <= 10:
  if i % 2 == 0:
    print(i)

  i += 1
```

```python
# print odd no between 0 to 10.

i = 1

while i <= 10:
  if i % 2 != 0:
    print(i)

  i += 1
```

```python
# print the sum of all the numbers between 1 to 10 (both inclusive)

# for loop

sum = 0

for i in range(1, 11):
  sum += i
  print(sum)
```

```python
# while loop
i = 1
sum = 0

while i <= 10:
  sum += i
  i += 1

print(sum)
```

```python
# print the multiplication of 5

i = 1
num = 5

while i <= 10:
  print(num, "x", i, "=", num*i)
  i += 1
```

```python
# Keep asking for a password until the correct one is entered.

p = ""
password = "password@123"

while p != password:
  p = input("Enter the password: ")

print("Correct password entered")
```

# Break

```python
for i in range(1, 11):
  if i == 5:
    break

  print(i)
```

```python
for i in range(1, 11):
  print(i)

  if i == 5:
    break
```

```python
# Keep asking for a password until the correct one is entered

password = "password@123"

while True:
  p = input("Enter the password: ")
  if p == password:
    break

print ("Outside the loop")
```

```python
# yaha per loop pura chalega kyuki 6 kabhi hoga hi nahi 2 - 2 ka gap hai

for i in range(1, 10, 2):
  if i == 6:
    break

  print(i)
```

```python
# yaha per condition True ho gaya and ruk gaya

for i in range(0, 10, 2):
  if i == 6:
    break

  print(i)
```

# Continue

```python
# isme 5 ko skip kar ke chelega

for i in range(1, 10):
  if i == 5:
    continue

  print(i)
```

```python
# isme koi coad hi nahi hai jo skip hoga isliye ye 1 to 9  chelaga

for i in range(1, 10):
  print(i)

  if i == 5:
    continue
```

```python
# print all the even numbers between 1 to 10 (both inclusive)

for i in range(1, 11):
  if i % 2 != 0:
    continue

  print(i)
```

# Nested loop

```python
for i in range(5):
  for j in range(3):
    print("i =", i, "j =", j)
```

```python
# print 1st row

for j in range(5):
  print("*", end=" ")
```

```python
for row in range(4):
  for col in range(5):
    print("*", end = " ")

  print()
```

```python
for row in range(1, 6):
  for col in range(1, 6):
    print(col, end = " ")

  print()
```

```python
for i in range(3):
  for j in range(1, 5):
    print(j*10, end = " ")

  print()
```

```python
for i in range(1, 6):
  for j in range(i):
    print("*", end = " ")

  print()
```

```python
for i in range(1, 6):
  for j in range(i):
    print(i, end = " ")

  print()
```

```python
for i in range(1, 6):
  for j in range(1, i+1):
    print(j, end = " ")

  print()
```

```python
n = 1

for row in range(1, 5):
  for col in range(row):
    print(n, end = " ")
    n = n+1

  print()
```

```python
# i print karne per same aata hai

for i in range(1, 4):
  for j in range(1, 4):
    print(i, end = " ")

  print()
```

```python
# j print karne per change hota hia

for i in range(1, 4):
  for j in range(1, 4):
    print(j, end = " ")

  print()
```
