```python
# Eligible to vote or not

age = int(input("Enter your age :"))

if age >= 18:
  print("You are eligible to vote")
else:
  print("You are not eligible to vote")
```

```python
# Check even or odd

a = int(input("Enter a number :"))

if a % 2 == 0:
  print("Even number")
else:
  print("Odd number")
```

```python
# identify number is positive , negative or zero.

n = int(input("Enter a number :"))

if(n > 0):
  print("Number is positive")
elif(n < 0):
  print("Number is negative")
else:
  print("Number is zero")
```

```python
# Takes two number, and print the larger number.
# if both are equal, print "Both are equal."

a = int(input("Input first number : "))
b = int(input("Input second number : "))

if a > b:
  print(a ,"is Larger")
elif a < b:
  print(b ,"is Larger")
else:
  print("Both are Equal")
```

```python
# Assign grades based on marks.

marks = int(input("Enter your marks :"))

if marks >= 90 :
  print("Grade A")
elif marks >= 75 :
  print("Grade B")
elif marks >= 60 :
  print("Grade C")
elif marks >= 40 :
  print("Grade D")
else:
  print("Try next time")
```

```python
# Body Mass Index (BMI) is a simple that helps detemine wheater a
# person's weight is appropariate for their height.


bmi = float(input("Enter your weight :"))

if bmi <= 18.5:
  print("Under Weight")
elif bmi <= 24.9:
  print("Normal Weight")
elif bmi <= 29.9:
  print("Over Weight")
else:
  print("Above obese")
```

```python
# Takses an integer a input and checks its divisibility

num = int(input("Enter a number :"))

if (num % 3 == 0) and (num % 5 == 0):
  print("Both divisible by both 3 and 5")
elif (num % 3 == 0):
  print("Divisible by 3")
elif (num % 5 == 0):
  print("Divisible by 5")
else:
  print("Not divisible by 3 or 5")
```
