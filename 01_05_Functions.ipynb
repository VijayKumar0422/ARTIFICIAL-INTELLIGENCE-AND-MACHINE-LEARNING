```python
def greet(name, age):
  print("name is", name)
  print("age is", age)

greet("vijay", 20)
```

```python
# yaha per ham ulta parameter dalte hia to ans galat aayega

def greet(name, age):
  print("name is", name)
  print("age is", age)

greet(20,"vijay")
```

```python
# Keyworded arguments
# yaha per parameter kaise bhe de ans sahi aayega

def greet(name, age):
  print("name is", name)
  print("age is", age)

greet(age = 20, name = "vijay")
```

```python
# return keyboard

def greet(a, b):
  return a + b

g = greet(2, 4)
print(g)
```

```python
# default argument
# isko ham sidha call kare show() aise to vijay print hoga
#usme jo likh kar kare to wah name print hoga

def show(name = "Vijay"):
  print(name)

show("hey")
```

```python
def total(numbers):
  s = 0

  for n in numbers:
    s += n

  return s

print(total([5, 10, 15]))
```

```python
def largest (a, b):
  if a > b:
    return a
  else:
    return b

print(largest(12, 25))
```
