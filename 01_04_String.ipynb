```python
# total character again and back are matched

word = "LEVEL"

same = 0

for i in range(len(word)):
  if word[i] == word[-(i + 1)]:
    same += 1
    print(i, -(i + 1), "---->", word[i], word[-(i+1)])

print(same)
```
