```python
marks = {
    "math" : 90,
    "science" : 80,
    "english" : 70
}

total = 0

for subject in marks.keys():
  total += marks[subject]

print(total / len(marks))
```

```python
employee = {
    "salary" : 50000
}

employee["salary"] = employee["salary"] * 1.10
print(employee["salary"])
```

```python
data = {
    "A" : 10,
    "B" : 20,
    "C" : 30
}

count = 0

for key in data.keys():
  if data[key] > 15:
    count += 1

print(count)
```
